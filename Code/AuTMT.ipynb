<h2 class="section-title">Overview</h2>
<div class="card">
<p class="lead">
The goal of this notebook is to provide a <b>reproducible TMT measurement workflow</b>.
It standardizes cranial CT orientation, automatically selects the most suitable axial slice,
and computes TMT on that slice. <b>No validation experiments are included</b> in this notebook.
</p>
</div>

</p>
